## Task 2 — Handle Missing Values

In this task, we identify missing values, calculate their percentages, visualize them, and handle them using suitable techniques.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load original dataset
df = pd.read_csv("penguins.csv")

# Create a copy for cleaning
cleaned_df = df.copy()

# Check missing values before cleaning
print("MISSING VALUES BEFORE CLEANING:")
print(cleaned_df.isnull().sum())

# Calculate total missing values and percentage
total_missing = cleaned_df.isnull().sum().sum()
total_cells = cleaned_df.size
missing_percentage = (total_missing / total_cells) * 100

print("\nTotal Missing Values:", total_missing)
print("Missing Percentage:", round(missing_percentage, 2), "%")

# Visualize missing values
missing_counts = cleaned_df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]

os.makedirs("outputs", exist_ok=True)

if not missing_counts.empty:
    plt.figure(figsize=(9, 5))
    sns.barplot(x=missing_counts.index, y=missing_counts.values)
    plt.title("Missing Values Per Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig("outputs/task2_missing_values.png")
    plt.show()
else:
    print("\nNo missing values found.")

# Fill numerical missing values with median
numerical_columns = cleaned_df.select_dtypes(
    include=np.number
).columns

for column in numerical_columns:
    cleaned_df[column] = cleaned_df[column].fillna(
        cleaned_df[column].median()
    )

# Fill categorical missing values with mode
categorical_columns = cleaned_df.select_dtypes(
    include="object"
).columns

for column in categorical_columns:
    mode_value = cleaned_df[column].mode()
    if not mode_value.empty:
        cleaned_df[column] = cleaned_df[column].fillna(
            mode_value[0]
        )

# Check missing values after cleaning
print("\nMISSING VALUES AFTER CLEANING:")
print(cleaned_df.isnull().sum())

print("\nTotal Missing Values After Cleaning:",
      cleaned_df.isnull().sum().sum())

# Save cleaned dataset
cleaned_df.to_csv("penguins_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

### Task 2 — Explanation

Missing values were identified using the isnull() function. Median values were used to fill missing numerical data because the median is less affected by outliers. Mode values were used for categorical data because they represent the most frequent category. Missing values were checked again after cleaning to verify the results.

## Task 3 — Remove Duplicate Records

In this task, we identify duplicate records, remove repeated rows, and verify the dataset after removal.

```python

import pandas as pd

# Load the cleaned dataset from Task 2
df = pd.read_csv("penguins_cleaned.csv")

# Check duplicate records
print("DUPLICATE RECORDS:")
print(df.duplicated())

# Count duplicates before removal
duplicates_before = df.duplicated().sum()
print("\nTotal Duplicates Before Removal:", duplicates_before)

# Remove duplicate records
df = df.drop_duplicates()

# Count duplicates after removal
duplicates_after = df.duplicated().sum()
print("\nTotal Duplicates After Removal:", duplicates_after)

# Display final dataset shape
print("\nDataset Shape After Removing Duplicates:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Save dataset
df.to_csv("penguins_no_duplicates.csv", index=False)

print("\nDataset saved successfully!")
```

## Task 4 — Standardize Categorical Data

In this task, we identify categorical columns, examine their unique values, and standardize text by removing extra spaces and converting values to lowercase.

```python

import pandas as pd

# Load dataset from Task 3
df = pd.read_csv("penguins_no_duplicates.csv")

# Create a copy
cleaned_df = df.copy()

# Identify categorical columns
categorical_columns = cleaned_df.select_dtypes(
    include=["object", "category", "string"]
).columns

print("Categorical Columns:")
print(list(categorical_columns))

# Display unique values before cleaning
print("\nUNIQUE VALUES BEFORE CLEANING:")
for column in categorical_columns:
    print(f"\n{column}:")
    print(cleaned_df[column].unique())

# Standardize categorical values
for column in categorical_columns:
    cleaned_df[column] = (
        cleaned_df[column]
        .astype("string")
        .str.strip()
        .str.lower()
    )

# Display unique values after cleaning
print("\nUNIQUE VALUES AFTER CLEANING:")
for column in categorical_columns:
    print(f"\n{column}:")
    print(cleaned_df[column].unique())

# Save standardized dataset
cleaned_df.to_csv("penguins_standardized.csv", index=False)

print("\nStandardized dataset saved successfully!")
```

### Task 4 — Explanation

Categorical columns were identified using select_dtypes(). The strip() function removed unnecessary spaces, while lower() converted text values into lowercase. This process helps maintain consistency and prevents different spellings or capitalization from being treated as separate categories.

## Task 5 — Detect and Review Outliers

In this task, we detect potential outliers in numerical features using boxplots and the Interquartile Range (IQR) method. We investigate unusual values before deciding whether to retain or remove them.

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load standardized dataset
df = pd.read_csv("penguins_standardized.csv")

# Select two numerical columns
columns = ["flipper_length_mm", "body_mass_g"]

# Create boxplots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for i, column in enumerate(columns):
    sns.boxplot(y=df[column], ax=axes[i])
    axes[i].set_title(f"Boxplot of {column}")
    axes[i].set_ylabel(column)

plt.tight_layout()
plt.savefig("outputs/task5_boxplots.png")
plt.show()

# Detect outliers using IQR
print("IQR OUTLIER ANALYSIS")

for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    print("Outlier Values:")
    print(outliers[["species", column]].to_string(index=False))

    print("Decision: Retain for further investigation.")

# Save reviewed dataset
df.to_csv("penguins_outliers_reviewed.csv", index=False)

print("\nOutlier analysis completed!")
print("Original observations have been retained.")
```

### Task 5 — Explanation

Outliers were investigated in flipper length and body mass using boxplots and the IQR method. The IQR method identifies values outside the lower and upper boundaries as potential outliers. These observations were retained because unusual measurements may represent natural differences between penguin species rather than data-entry errors.

## Task 8 — Encoding Categorical Data

In this task, categorical features are converted into numerical form using Label Encoding and One-Hot Encoding.


```python

import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load feature-engineered dataset
df = pd.read_csv("penguins_feature_engineered.csv")

# Identify categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "category", "string"]
).columns

print("Categorical Columns:")
print(list(categorical_columns))

# Display data before encoding
print("\nDATASET BEFORE ENCODING:")
print(df[["species", "island", "sex"]].head())

# Label Encoding for sex
label_encoder = LabelEncoder()
df["sex_encoded"] = label_encoder.fit_transform(df["sex"])

print("\nLABEL ENCODING MAPPING:")
for number, category in enumerate(label_encoder.classes_):
    print(category, "=", number)

# One-Hot Encoding for island
df = pd.get_dummies(df, columns=["island"], dtype=int)

# Display data after encoding
print("\nDATASET AFTER ENCODING:")
print(df.head())

# Save encoded dataset
df.to_csv("penguins_encoded.csv", index=False)

print("\nEncoding completed successfully!")
print("Encoded dataset saved as penguins_encoded.csv")

```

### Task 8 — Explanation

Label Encoding was applied to the sex column to convert categories into numerical values. One-Hot Encoding was applied to the island column to create separate binary columns for each island. These techniques convert categorical information into a numerical format suitable for machine learning algorithms. The species column was kept unchanged because it may be used as the target variable.

## Task 9 — Feature Scaling

In this task, numerical features are scaled using Min-Max Scaling and Standardization. These techniques help bring numerical features to comparable scales.

```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Load encoded dataset
df = pd.read_csv("penguins_encoded.csv")

# Select numerical features
features = ["bill_length_mm", "body_mass_g"]

# Keep original values for comparison
comparison = df[features].copy()

# Apply Min-Max Scaling
minmax_scaler = MinMaxScaler()
minmax_values = minmax_scaler.fit_transform(df[features])

# Apply Standardization
standard_scaler = StandardScaler()
standard_values = standard_scaler.fit_transform(df[features])

# Add scaled columns
for i, column in enumerate(features):
    df[column + "_minmax"] = minmax_values[:, i]
    df[column + "_standardized"] = standard_values[:, i]

# Display before and after values
comparison["bill_length_minmax"] = df["bill_length_mm_minmax"]
comparison["bill_length_standardized"] = df["bill_length_mm_standardized"]
comparison["body_mass_minmax"] = df["body_mass_g_minmax"]
comparison["body_mass_standardized"] = df["body_mass_g_standardized"]

print("BEFORE AND AFTER SCALING:")
print(comparison.head(10).round(3))

# Save scaled dataset
df.to_csv("penguins_scaled.csv", index=False)

print("\nFeature scaling completed successfully!")
print("Scaled dataset saved as penguins_scaled.csv")
```

### Task 9 — Explanation

Feature scaling was applied to bill length and body mass using Min-Max Scaling and Standardization. Min-Max Scaling converts values into a range between 0 and 1, while Standardization transforms values so they have a mean of approximately 0 and standard deviation approximately 1. These techniques help machine learning algorithms work with numerical features that have different scales.

# Section 6 — Additional Questions

### Q1. Why is data preprocessing important before training an AI model?

Data preprocessing is important because raw data may contain missing values, duplicate records, incorrect formats, and inconsistent information. Preprocessing cleans and organizes the data before training. It improves data quality and helps the AI model learn meaningful patterns and produce more reliable predictions.

### Q2. What is the difference between missing values and duplicate records?

Missing values are empty or unavailable entries in a dataset. For example, a penguin record may have no value for body mass.

Duplicate records are repeated rows containing the same information more than once. Missing values represent incomplete information, while duplicate records represent repeated information.

### Q3. Why is the median sometimes preferred over the mean when handling missing values?

The median is sometimes preferred because it is less affected by extremely high or low values (outliers). The mean can change significantly when outliers are present, while the median usually remains more stable. Therefore, the median can provide a better replacement for missing numerical values in data containing outliers.

### Q4. What are outliers, and why should we not automatically remove them?

Outliers are values that are unusually different from most other values in a dataset. They may occur because of measurement errors, data-entry mistakes, or natural variations. We should not automatically remove them because some outliers may contain important information. Removing valid observations could result in losing useful data and affecting model performance.

### Q5. What is the difference between Label Encoding and One-Hot Encoding?

Label Encoding converts each category into a numerical value. For example, female can be represented as 0 and male as 1.

One-Hot Encoding creates separate binary columns for different categories. Each column contains 0 or 1 to represent whether a record belongs to that category.

Label Encoding uses numerical labels, while One-Hot Encoding creates separate columns for categories.

### Q6. Why is feature scaling important for some machine learning algorithms?

Feature scaling is important because numerical features may have very different ranges. For example, body mass may be measured in thousands of grams, while another feature may contain much smaller values. Scaling brings features to comparable ranges. It helps algorithms such as K-Nearest Neighbors (KNN), Support Vector Machines (SVM), and gradient-based models work more effectively.

### Q7. What problems could occur if we train a machine learning model on uncleaned data?

Training a machine learning model on uncleaned data can cause several problems:

1. Missing values may cause errors or prevent the model from learning properly.
2. Duplicate records may give repeated observations too much influence.
3. Incorrect or inconsistent data may confuse the model.
4. Extreme values may affect model predictions.
5. Unencoded categorical data may not be accepted by some algorithms.
6. Unscaled features may cause problems for algorithms that depend on distances or feature magnitudes.

As a result, the model may learn incorrect patterns and produce unreliable predictions.
